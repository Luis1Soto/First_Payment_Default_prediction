### Data Processing 

we eliminate features that are filled with a single value since it weill not be useful for models

```python
constant_cols = [col for col in Data.columns if Data[col].nunique() == 1]
Data = Data.drop(columns=constant_cols)
print("Constant columns eliminated:", constant_cols)
```

```python
for col in Data.columns[Data.isnull().sum() > 0]:
    plt.figure(figsize=(8, 4))
    sns.histplot(Data[col].dropna(), kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
knn_imputer = KNNImputer(n_neighbors=5)

Data_imputed_knn = pd.DataFrame(knn_imputer.fit_transform(Data), columns=Data.columns)

print(Data_imputed_knn.isnull().sum())
```
